```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('student_performance_dirty.csv')

# Fill missing values with median
data['Hours Studied'] = data['Hours Studied'].fillna(data['Hours Studied'].median())
data['Sleep Hours'] = data['Sleep Hours'].fillna(data['Sleep Hours'].median())

# Show number of empty cells in each column
empty_cells = data.isnull().sum()
print(empty_cells)


# Add Rupy column starting from 1000
data['Rupy'] = range(1000, 1000 + len(data))
# Convert values into k format
data['Rupy'] = (data['Rupy']).astype(str) + 'k'
data


```
