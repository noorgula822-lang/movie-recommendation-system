```python
import numpy as num
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Function to filter total_bill values
def filter_bill(data):
    filtered_data = data[
        (data['total_bill'] > 10) &
        (data['total_bill'] < 30)
    ]
    return filtered_data

# Call function
result = filter_bill(data)

print(result)
# print(data[data['smoker'] == 'Yes'].shape[0])
# new = data['smoker'] == 'No'
# data[new]
# type(data['smoker'])
# data['smoker']
# data[['tip','day','time']]
# data.iloc[[1,10]]
# data.iloc[:,[1,2]]
# data.head()
# Function to count total smokers
def total_smokers(data):
    smokers = data[data['smoker'] == 'No']
    return smokers.shape[0]

# Call function
result = total_smokers(data)

print("Total Smokers:", result)
```

```python
# sns.histplot(data=data,x='total_bill',bins=5)
ax=sns.histplot(data=data,x='smoker',kde=True,color='green',hue='sex')
# Dark background
plt.style.use('dark_background')
            # Different colors for bins
colors = ['red', 'blue', 'green', 'orange', 'purple',
          'pink', 'brown', 'yellow', 'cyan', 'black']

# Apply colors
for patch, color in zip(ax.patches, colors):
    patch.set_facecolor(color)
```
