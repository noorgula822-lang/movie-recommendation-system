```python
import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import LabelEncoder

from sklearn.ensemble import RandomForestClassifier

from sklearn.linear_model import LinearRegression

from sklearn.metrics import accuracy_score, r2_score

from flask import Flask, request



# =========================
# Load Dataset
# =========================

data = pd.read_csv(
    "StudentsPerformance.csv"
)


print(data.head())

print(data.columns)



# =========================
# Encode Gender
# =========================

encoder = LabelEncoder()


data['gender_encoded'] = encoder.fit_transform(
    data['gender']
)



# =========================
# Input Feature
# =========================

X = data[['writing score']]



# =========================
# Gender Model
# =========================

y_gender = data['gender_encoded']


X_train, X_test, y_train, y_test = train_test_split(

    X,

    y_gender,

    test_size=0.2,

    random_state=42

)



gender_model = RandomForestClassifier()


gender_model.fit(

    X_train,

    y_train

)



gender_pred = gender_model.predict(

    X_test

)



print(
"Gender Accuracy:",
accuracy_score(
    y_test,
    gender_pred
)

)





# =========================
# Math Score Model
# =========================


y_math = data['math score']



X_train2, X_test2, y_train2, y_test2 = train_test_split(

    X,

    y_math,

    test_size=0.2,

    random_state=42

)



math_model = LinearRegression()



math_model.fit(

    X_train2,

    y_train2

)



math_pred = math_model.predict(

    X_test2

)



print(

"Math R2 Score:",

r2_score(

    y_test2,

    math_pred

)

)





# =========================
# Prediction Function
# =========================


def predict_student(writing_score):


    new_data = pd.DataFrame(

        [[writing_score]],

        columns=['writing score']

    )



    gender_prediction = gender_model.predict(

        new_data

    )



    math_prediction = math_model.predict(

        new_data

    )



    gender = encoder.inverse_transform(

        gender_prediction

    )[0]



    math = round(

        math_prediction[0],

        2

    )



    return gender, math





# Test Prediction

print(
predict_student(80)
)





# =========================
# Flask Website
# =========================


app = Flask(__name__)




@app.route("/", methods=["GET","POST"])

def home():


    result = ""



    if request.method == "POST":


        writing = float(

            request.form["writing"]

        )



        gender, math = predict_student(

            writing

        )



        result = f"""

        <h2>

        Predicted Gender: {gender}

        </h2>


        <h2>

        Predicted Math Score: {math}

        </h2>

        """



    return f"""


    <html>


    <head>


    <title>
    Student AI Prediction
    </title>


    <style>


    body{{

    font-family:Arial;

    background:#f2f2f2;

    text-align:center;

    padding:50px;

    }}


    .card{{

    background:white;

    width:400px;

    margin:auto;

    padding:30px;

    border-radius:15px;

    box-shadow:0px 0px 15px gray;

    }}



    input{{

    padding:12px;

    width:80%;

    }}



    button{{

    padding:12px 25px;

    background:blue;

    color:white;

    border:none;

    }}


    </style>


    </head>



    <body>



    <div class="card">


    <h1>

    Student Performance AI

    </h1>


    <form method="POST">


    <input

    type="number"

    name="writing"

    placeholder="Enter Writing Score"

    required>



    <br><br>


    <button>

    Predict

    </button>


    </form>


    {result}


    </div>



    </body>


    </html>


    """



app.run(

    host="127.0.0.1",

    port=5000

)
```
