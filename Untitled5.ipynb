```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv("StudentsPerformance.csv")   # Change filename if needed

# Calculate average math score for each gender
avg_math = data.groupby("gender")["math score"].mean()

# Create bar chart
plt.figure(figsize=(6,5))
plt.bar(avg_math.index, avg_math.values)

# Add title and labels
plt.title("Average Math Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Math Score")

# Show values on bars
for i, value in enumerate(avg_math.values):
    plt.text(i, value + 0.5, f"{value:.1f}", ha='center')

# Display chart
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv("StudentsPerformance.csv")   # Change filename if needed

# Calculate average math score for each gender
avg_math = data.groupby("gender")["math score"].mean()

# Create pie chart
plt.figure(figsize=(6,6))
plt.pie(
    avg_math,
    labels=avg_math.index,
    autopct='%1.1f%%',
    startangle=90
)

# Add title
plt.title("Average Math Score by Gender")

# Display chart
plt.show()
```

```python
import pandas as pd

# ==============================
# Load Dataset
# ==============================
data = pd.read_csv("sale_order.csv")

# ==============================
# Remove Completely Empty Columns
# ==============================
data = data.dropna(axis=1, how="all")

# ==============================
# Calculate Missing Percentage
# ==============================
missing_percent = (data.isnull().sum() / len(data)) * 100

# Show only columns that have missing values
print("Missing Percentage:")
print(missing_percent[missing_percent > 0])

# ==============================
# Lists to Store Results
# ==============================
filled_columns = []
dropped_columns = []

# ==============================
# Handle Missing Values
# ==============================
for col in data.columns:

    pct = missing_percent[col]

    # No Missing Values
    if pct == 0:
        continue

    # 0–5% Missing → Fill
    elif pct <= 5:

        if data[col].dtype == "object":
            data[col] = data[col].fillna(data[col].mode()[0])
        else:
            data[col] = data[col].fillna(data[col].median())

        filled_columns.append(col)

    # 5–30% Missing → Fill
    elif pct <= 30:

        if data[col].dtype == "object":
            data[col] = data[col].fillna(data[col].mode()[0])
        else:
            data[col] = data[col].fillna(data[col].median())

        filled_columns.append(col)

    # 30–60% Missing → Drop
    elif pct <= 60:

        data = data.drop(columns=[col])
        dropped_columns.append(col)

    # More than 60% Missing → Drop
    else:

        data = data.drop(columns=[col])
        dropped_columns.append(col)

# ==============================
# Final Results
# ==============================
print("\nFilled Columns:")
print(filled_columns)

print("\nDropped Columns:")
print(dropped_columns)

print("\nRemaining Columns:", data.shape[1])
print("Dataset Shape:", data.shape)

# ==============================
# Save Cleaned Dataset
# ==============================
data.to_csv("sale_order_cleaned.csv", index=False)

print("\nCleaned dataset saved as 'sale_order_cleaned.csv'")
```
