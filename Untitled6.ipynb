```python
# ============================================
# 1. Imports
# ============================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error,
    r2_score, mean_absolute_percentage_error
)

from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor

import joblib
import warnings
warnings.filterwarnings("ignore")

# ============================================
# 2. Load Data
# ============================================
data = pd.read_csv("Housing.csv")
print("Shape:", data.shape)
data.head()

# ============================================
# 3. Explore Categorical Columns
# ============================================
categorical_columns = data.select_dtypes(include="object").columns.tolist()

for col in categorical_columns:
    print(f"\n{col}")
    print(data[col].unique())
    print(data[col].value_counts())

# ============================================
# 4. Correlation Heatmap (numeric columns, BEFORE encoding)
# ============================================
plt.figure(figsize=(10, 8))
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap (Raw Numeric Columns)")
plt.show()

# ============================================
# 5. Distributions
# ============================================
data.hist(figsize=(15, 10))
plt.tight_layout()
plt.show()

numerical_columns = data.select_dtypes(include=["int64", "float64"]).columns.tolist()

for col in numerical_columns:
    plt.figure(figsize=(6, 3))
    sns.histplot(data[col], kde=True)
    plt.title(f"Distribution: {col}")
    plt.show()

for col in numerical_columns:
    plt.figure(figsize=(6, 3))
    sns.boxplot(x=data[col])
    plt.title(f"Boxplot: {col}")
    plt.show()

for col in numerical_columns:
    if col != "price":
        plt.figure(figsize=(6, 4))
        sns.scatterplot(x=data[col], y=data["price"])
        plt.title(f"{col} vs Price")
        plt.show()

# ============================================
# 6. Outlier Detection + Handling (IQR clipping) — run ONCE
# ============================================
for col in numerical_columns:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = data[(data[col] < lower) | (data[col] > upper)]
    print(f"{col}: {len(outliers)} outliers before clipping")

    data[col] = data[col].clip(lower, upper)

# ============================================
# 7. Encode ALL Categorical Columns — do this ONCE, fully, before feature engineering
# ============================================
binary_cols = ['mainroad', 'guestroom', 'basement', 'hotwaterheating', 'airconditioning', 'prefarea']

le = LabelEncoder()
for col in binary_cols:
    data[col] = le.fit_transform(data[col])   # yes -> 1, no -> 0

data = pd.get_dummies(data, columns=['furnishingstatus'], drop_first=True)

# Confirm nothing text-based remains
print("Object columns remaining:", data.select_dtypes(include='object').columns.tolist())
data.head()

# ============================================
# 8. Feature Engineering — now safe, since binary_cols are already 0/1 numeric
# ============================================
data["total_rooms"] = data["bedrooms"] + data["bathrooms"]
data["area_per_bedroom"] = data["area"] / data["bedrooms"].replace(0, 1)

data["luxury_house"] = (
    (data["airconditioning"] == 1) & (data["mainroad"] == 1)
).astype(int)

data["total_facilities"] = (
    data["guestroom"] + data["basement"] + data["hotwaterheating"] +
    data["airconditioning"] + data["prefarea"]
)

data.head()

# ============================================
# 9. Correlation Heatmap — AFTER encoding + feature engineering
# ============================================
correlation = data.corr(numeric_only=True)
print(correlation["price"].sort_values(ascending=False))

plt.figure(figsize=(12, 10))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap (Final Feature Set)")
plt.show()

# ============================================
# 10. Feature Importance (quick check before modeling)
# ============================================
X = data.drop("price", axis=1)
y = data["price"]

print("Features Shape:", X.shape)
print("Target Shape  :", y.shape)

rf_check = RandomForestRegressor(random_state=42)
rf_check.fit(X, y)

importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": rf_check.feature_importances_
}).sort_values(by="Importance", ascending=False)

print(importance)

importance.plot(x="Feature", y="Importance", kind="bar", figsize=(12, 6))
plt.title("Feature Importance (Random Forest)")
plt.ylabel("Importance")
plt.tight_layout()
plt.show()

# ============================================
# 11. Train/Test Split
# ============================================
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("X_train Shape:", X_train.shape)
print("X_test Shape :", X_test.shape)
print("Training %   :", round(len(X_train) / len(X) * 100, 2), "%")
print("Testing %    :", round(len(X_test) / len(X) * 100, 2), "%")

# ============================================
# 12. Scaling — fit on train, apply to both. Used consistently for ALL models
#     (scaling doesn't hurt tree models, and keeps everything uniform/correct)
# ============================================
scaler = StandardScaler()
X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train), columns=X_train.columns, index=X_train.index)
X_test_scaled = pd.DataFrame(scaler.transform(X_test), columns=X_test.columns, index=X_test.index)

# ============================================
# 13. Train ALL Models (on the same scaled data, so results are comparable)
# ============================================
models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(n_estimators=100, random_state=42),
    "Gradient Boosting": GradientBoostingRegressor(random_state=42),
    "XGBoost": XGBRegressor(n_estimators=100, learning_rate=0.1, max_depth=5, random_state=42),
    "LightGBM": LGBMRegressor(random_state=42),
    "CatBoost": CatBoostRegressor(verbose=0, random_state=42),
}

predictions = {}

for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    predictions[name] = model.predict(X_test_scaled)
    print(f"{name} trained successfully.")

# ============================================
# 14. Evaluate ALL Models
# ============================================
def evaluate_model(model_name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    mape = mean_absolute_percentage_error(y_true, y_pred)
    return {"Model": model_name, "MAE": mae, "MSE": mse, "RMSE": rmse, "R2 Score": r2, "MAPE": mape}

results = pd.DataFrame([
    evaluate_model(name, y_test, y_pred) for name, y_pred in predictions.items()
])

print("\nAll Model Results (sorted by R2 Score):")
results_sorted = results.sort_values(by="R2 Score", ascending=False).reset_index(drop=True)
print(results_sorted)

print("\nAll Model Results (sorted by RMSE):")
print(results.sort_values(by="RMSE", ascending=True).reset_index(drop=True))

# ============================================
# 15. Identify and Confirm the Best Model
# ============================================
best_row = results_sorted.iloc[0]
best_model_name = best_row["Model"]
best_model = models[best_model_name]
y_pred_best = predictions[best_model_name]

print(f"\nBest Model: {best_model_name}")
print(best_row)

# ============================================
# 16. Actual vs Predicted — Best Model
# ============================================
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred_best, alpha=0.6)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')  # perfect-prediction line
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title(f"Actual vs Predicted Price ({best_model_name})")
plt.show()

comparison_best = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred_best
})
print(comparison_best.head(10))

# ============================================
# 17. Hyperparameter Tuning (Random Forest example — optional but wired correctly)
# ============================================
param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

grid_search = GridSearchCV(
    estimator=RandomForestRegressor(random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring='r2',
    n_jobs=-1,
    verbose=1
)
grid_search.fit(X_train_scaled, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best CV R2 Score:", grid_search.best_score_)

# Evaluate the tuned model on the actual test set
tuned_rf = grid_search.best_estimator_
y_pred_tuned = tuned_rf.predict(X_test_scaled)
tuned_result = evaluate_model("Random Forest (Tuned)", y_test, y_pred_tuned)
print("\nTuned Random Forest Test Performance:")
print(tuned_result)

# If the tuned model beats the current best, use it as the final model
if tuned_result["R2 Score"] > best_row["R2 Score"]:
    print("\nTuned Random Forest outperforms the previous best model — using it as final.")
    best_model = tuned_rf
    best_model_name = "Random Forest (Tuned)"
    y_pred_best = y_pred_tuned

# ============================================
# 18. Save the Final Best Model
# ============================================
joblib.dump(best_model, "best_housing_model.pkl")
joblib.dump(scaler, "scaler.pkl")

print(f"\nFinal model saved: {best_model_name} -> best_housing_model.pkl")
print("Scaler saved -> scaler.pkl")
```
